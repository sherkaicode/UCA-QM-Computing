In classicial bits = {0, 1}

In Quantum State: Superposition of 
$$ \ket{\Psi} = a \ket{0} + b \ket{1}$$
where $a, b \in \mathbb{C}$, and $|a|^2 + |b|^2 = 1$

$$
\ket{\Psi} = \begin{pmatrix} a \\ b \end{pmatrix} = a \begin{pmatrix} 1 \\ 0 \end{pmatrix} + b \begin{pmatrix} 0 \\ 1\end{pmatrix}
$$

```python
# There is a library in qiskit
from qiskit import quantum_info as qi

num_qubits = 2

sv = qi.Statevector(np.random.randint(0, 2, size=2**num_qubits)) # ket vector
```

```python
from qiskit.visualization import array_to_latex
array_to_latex(sv.data, prefix="\\text{Statevector} = ")
```
